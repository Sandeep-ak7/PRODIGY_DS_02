```python
df['Survived'].value_counts()
```

```python
sns.countplot(x='Survived',data=df,palette='viridis',)
plt.xlabel("Survival Status Of people")
plt.ylabel("Number of people")
plt.xticks(ticks=[0,1],labels=['Not Survived','Survived'])
plt.show()                    
```

```python
plt.pie(df['Survived'].value_counts(),explode=[0,0.03],autopct="%1.2f%%",labels=['Not survived','Survived'])
plt.title("Survival of people")
plt.show()
```

```python
sns.countplot(x='Sex',data=df,palette='viridis',)
plt.xlabel("Gender")
plt.ylabel("Number of people")
plt.show()
```

```python
sns.countplot(x='Survived',hue='Sex',data=df,palette='magma',)
plt.xlabel("Survival status")
plt.ylabel("Number of people")
plt.xticks(ticks=[0,1],labels=['Not survived','survived'])
plt.show()
```

```python
df[df['Sex'] == 'male'].Survived.groupby(df.Survived).count().plot(kind='pie',
figsize=(3, 6),explode=[0,0.05],autopct='%1.1f%%',labels=["Not survived","Survived"])
plt.ylabel("")
plt.title("Male survival rate")
plt.show()
```

```python
sns.histplot(x='Fare',hue='Survived',data=df,kde=True)
plt.legend(labels=['survived','not survived'])
plt.show()
```

```python
sns.countplot(x ='Survived', hue='SibSp',data=df,palette='plasma')
plt.xticks(ticks=[0,1],labels=['Not survived','survived'])
plt.xlabel("Survival status")
plt.ylabel("Number of people")
plt.title("Survival population of Sibsp")
plt.show()

sns.countplot(x ='Survived',hue='Parch',data=df,palette='inferno')
plt.xticks(ticks=[0,1],labels=['Not survived','survived'])
plt.title("Survival population of Parch")
plt.xlabel("Survival status")
plt.ylabel("Number of people")
plt.show()
```

```python
fig,axes = plt.subplots(1, 2, figsize=(12, 6))
sns.countplot(x='SibSp',data=df,ax=axes[0],palette='BuPu')
sns.countplot(x='Parch',data=df,ax=axes[1],palette='BuPu')
plt.show()
```

```python
sns.histplot(x='Age',hue='Survived',data=df,kde=True)
plt.legend(labels=['survived','not survived'])
plt.show()
```
